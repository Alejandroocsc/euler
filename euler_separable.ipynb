# Enfriamiento de una taza de café

Una taza de café comienza a 80 °C en una habitación a 20 °C. Modelamos su temperatura $T(t)$, en grados Celsius, mediante

$$\frac{dT}{dt}=-0.5(T-20),\qquad T(0)=80.$$

Aquí $t$ está en minutos. Compararemos la solución exacta con el método de Euler en $[0,1]$ usando $h=0.2$.

## Solución por separación de variables

Como la temperatura inicial es mayor que la del ambiente, separamos e integramos:

$$\frac{dT}{T-20}=-0.5\,dt, \qquad \ln|T-20|=-0.5t+C.$$

Al despejar, $T-20=Ce^{-0.5t}$. La condición $T(0)=80$ da $C=60$, así que

$$\boxed{T(t)=20+60e^{-0.5t}}.$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Aproximación con Euler

Partimos de $T_0=80$ y aplicamos $T_{i+1}=T_i+h\,[-0.5(T_i-20)]$ durante cinco pasos.

```python
def f(t, temperatura):
    return -0.5 * (temperatura - 20)

t0, tf, h = 0.0, 1.0, 0.2
numero_pasos = round((tf - t0) / h)
tiempos = np.linspace(t0, tf, numero_pasos + 1)
temperaturas = np.zeros(numero_pasos + 1)
temperaturas[0] = 80.0

# Usamos la pendiente actual para estimar la temperatura siguiente.
for i in range(numero_pasos):
    temperaturas[i + 1] = temperaturas[i] + h * f(tiempos[i], temperaturas[i])

exactas = 20 + 60 * np.exp(-0.5 * tiempos)
errores = np.abs(exactas - temperaturas)
```

```python
print('Tiempo (min)   Euler (°C)   Exacta (°C)   Error absoluto (°C)')
for tiempo, euler, exacta, error in zip(tiempos, temperaturas, exactas, errores):
    print(f'{tiempo:>12.1f} {euler:>12.6f} {exacta:>13.6f} {error:>21.6f}')
```

```python
# Una curva suave ayuda a comparar la solución exacta con los puntos de Euler.
t_curva = np.linspace(t0, tf, 200)
plt.figure(figsize=(8, 5))
plt.plot(t_curva, 20 + 60 * np.exp(-0.5 * t_curva), label='Solución exacta')
plt.plot(tiempos, temperaturas, 'o--', label='Euler (h = 0.2)')
plt.title('Enfriamiento de una taza de café')
plt.xlabel('Tiempo (min)')
plt.ylabel('Temperatura (°C)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
